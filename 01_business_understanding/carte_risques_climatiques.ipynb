```python
# Création du notebook d'analyse des risques
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

# Données d'exemple pour la carte des risques
risks_data = {
    'Risk_Type': ['Sécheresse', 'Gel', 'Chaleur Extrême', 'Inondation', 'Vent Violent'],
    'Impact_Agriculture': [9, 8, 7, 6, 5],
    'Frequence_Projection': [8, 6, 7, 5, 4],
    'Urgence_Action': [9, 7, 8, 6, 5]
}

df_risks = pd.DataFrame(risks_data)

# Visualisation
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Impact sur l'agriculture
axes[0,0].barh(df_risks['Risk_Type'], df_risks['Impact_Agriculture'])
axes[0,0].set_title('Impact sur l Agriculture')
axes[0,0].set_xlabel('Score Impact (1-10)')

# 2. Fréquence projetée
axes[0,1].barh(df_risks['Risk_Type'], df_risks['Frequence_Projection'], color='orange')
axes[0,1].set_title('Fréquence Projetée')
axes[0,1].set_xlabel('Score Fréquence (1-10)')

# 3. Urgence d'action
axes[1,0].barh(df_risks['Risk_Type'], df_risks['Urgence_Action'], color='red')
axes[1,0].set_title('Urgence d Action')
axes[1,0].set_xlabel('Score Urgence (1-10)')

# 4. Matrice risque
scatter = axes[1,1].scatter(df_risks['Impact_Agriculture'], 
                           df_risks['Frequence_Projection'],
                           s=df_risks['Urgence_Action']*100,
                           alpha=0.6)
axes[1,1].set_xlabel('Impact Agriculture')
axes[1,1].set_ylabel('Fréquence Projection')
axes[1,1].set_title('Matrice Risques Climatiques')

# Ajout annotations
for i, risk in enumerate(df_risks['Risk_Type']):
    axes[1,1].annotate(risk, 
                      (df_risks['Impact_Agriculture'][i], 
                       df_risks['Frequence_Projection'][i]),
                      xytext=(5, 5), textcoords='offset points')

plt.tight_layout()
plt.savefig('matrice_risques_climatiques.png', dpi=300, bbox_inches='tight')
plt.show()

print("🎯 Carte des risques climatiques générée avec succès!")
```
